# Calculo da Folha de Pagamento com bonificações por funcionário

#### 50% das despesas com plano de saúde serão arcadas pela empresa
#### 8% do salário mínimo de bonificação 
#### 5% de bônus em cima do salário Bruto

##### Desafio: Implementar um programa que calcule o salário líquido de cada funcionário

```python
funcionarios = [
    {"nome": "Funcionario A", "salarioBruto": 2000, "descontoPlanoSaude" : 100, "descontosDiversos" : 300},
    {"nome": "Funcionario B", "salarioBruto": 2500, "descontoPlanoSaude" : 125, "descontosDiversos" : 375},
    {"nome": "Funcionario C", "salarioBruto": 3000, "descontoPlanoSaude" : 150, "descontosDiversos" : 450},
    {"nome": "Funcionario D", "salarioBruto": 4000, "descontoPlanoSaude" : 200, "descontosDiversos" : 600},
    {"nome": "Funcionario E", "salarioBruto": 4500, "descontoPlanoSaude" : 225, "descontosDiversos" : 675},
]
```

```python
salarioMinimo = 1500
```

```python
def calculoNovoSalario(salarioBruto,descontoPlanoSaude,descontosDiversos):
    salarioLiquido = salarioBruto
    salarioLiquido -= descontosDiversos
    salarioLiquido -= descontoPlanoSaude * (50/100)
    salarioLiquido += salarioMinimo * (8/100)
    salarioLiquido += salarioBruto * (5/100)

    return salarioLiquido
```

```python
for funcionario in funcionarios:
    novoSalario = calculoNovoSalario(
        funcionario["salarioBruto"],
        funcionario["descontoPlanoSaude"],
        funcionario["descontosDiversos"]
    )
    print(f'{funcionario["nome"]} - Salário Líquido: R${novoSalario:.2f}')
```
